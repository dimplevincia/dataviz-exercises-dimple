```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2000 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE
# Task 1 — Scatter: GDP per capita vs Life Expectancy

# Filter one continent (example: Europe)
europe_df = df.loc[df['continent'] == 'Europe']

# Filter only 2000 and 2007
scatter_df = europe_df[
    europe_df['year'].isin([2000, 2007])
]

# Create scatter plot
fig = px.scatter(
    scatter_df,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    hover_name='country',
    size='pop',
    template='simple_white',
    color_discrete_sequence=['#BBBBBB', 'crimson']
)

# Layout styling
fig.update_layout(
    title='European countries improved life expectancy as GDP per capita increased',
    font=dict(family='Arial', size=12),
    width=950,
    height=600,
    showlegend=True
)

# Remove clutter
fig.update_xaxes(
    title='GDP per Capita',
    showgrid=False
)

fig.update_yaxes(
    title='Life Expectancy',
    gridcolor='#EEEEEE'
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE
# Task 2 — Bubble Chart: GDP vs Life Expectancy (2007)

# Filter 2007 data
bubble_df = df[df['year'] == 2007]

# Choose continent to highlight
highlight_continent = 'Asia'

# Create colour column
bubble_df['highlight'] = bubble_df['continent'].apply(
    lambda x: highlight_continent if x == highlight_continent else 'Other'
)

# Create bubble chart
fig = px.scatter(
    bubble_df,
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='highlight',
    hover_name='country',
    log_x=True,
    template='simple_white',
    color_discrete_map={
        highlight_continent: 'crimson',
        'Other': '#DDDDDD'
    }
)

# Annotation
fig.add_annotation(
    x=30000,
    y=82,
    text='Asian countries combined rapid economic growth with longer life expectancy',
    showarrow=True,
    arrowhead=2,
    font=dict(size=11, color='crimson')
)

# Layout styling
fig.update_layout(
    title='Asian countries showed strong gains in both wealth and life expectancy by 2007',
    font=dict(family='Arial', size=12),
    width=1000,
    height=650,
    showlegend=False
)

# Clean styling
fig.update_xaxes(
    title='GDP per Capita (Log Scale)',
    showgrid=False
)

fig.update_yaxes(
    title='Life Expectancy',
    gridcolor='#EEEEEE'
)

fig.show()
```
